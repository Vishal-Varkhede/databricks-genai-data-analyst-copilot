```python
# ================================================================
# CELL 10 — API REQUEST VALIDATION
# ================================================================

print("=" * 70)
print("API REQUEST VALIDATION")
print("=" * 70)

# ------------------------------------------------
# Example request
# ------------------------------------------------

example_request = {
    "question": "Which region generated the highest revenue?"
}

print()
print("Example request:")
print(
    json.dumps(
        example_request,
        indent=2
    )
)

# ------------------------------------------------
# Validate request
# ------------------------------------------------

request_validation = validate_api_request(
    example_request
)

print()
print("Validation result:")
print(
    json.dumps(
        request_validation,
        indent=2,
        default=str
    )
)

if not request_validation.get(
    "valid",
    False
):

    raise RuntimeError(
        "API request validation failed: "
        + str(
            request_validation.get(
                "error"
            )
        )
    )

print()
print("API request validation: PASS")
```

```python
# ================================================================
# CELL 11 — API REQUEST EXECUTION
# ================================================================

print("=" * 70)
print("API REQUEST EXECUTION")
print("=" * 70)

api_response = build_api_response(
    example_request
)

print()
print("API response:")
print(
    json.dumps(
        api_response,
        indent=2,
        default=str
    )
)

if not api_response.get(
    "success",
    False
):

    raise RuntimeError(
        "API execution failed: "
        + str(
            api_response.get(
                "error"
            )
        )
    )

print()
print("API request execution: PASS")
```

```python
# ================================================================
# CELL 12 — API RESPONSE VALIDATION
# ================================================================

print("=" * 70)
print("API RESPONSE VALIDATION")
print("=" * 70)

required_response_fields = [
    "success",
    "question",
    "route",
    "answer",
    "sql",
    "data",
    "sources",
    "error"
]

missing_fields = []

for field in required_response_fields:

    if field not in api_response:

        missing_fields.append(
            field
        )

    print(
        f"{'PASS' if field in api_response else 'FAIL'} - "
        f"{field}"
    )

print()

if missing_fields:

    raise RuntimeError(
        "Missing response fields: "
        + ", ".join(
            missing_fields
        )
    )

print(
    "Response structure validation: PASS"
)
```

```python
# ================================================================
# CELL 13 — ROUTE VALIDATION
# ================================================================

print("=" * 70)
print("ROUTE VALIDATION")
print("=" * 70)

valid_routes = [
    "sql",
    "rag",
    "hybrid"
]

actual_route = api_response.get(
    "route"
)

print(
    "Detected route:",
    actual_route
)

if actual_route not in valid_routes:

    raise RuntimeError(
        "Invalid route returned: "
        + str(actual_route)
    )

print()
print(
    "Route validation: PASS"
)
```

```python
# ================================================================
# CELL 14 — DEPLOYMENT READINESS
# ================================================================

print("=" * 70)
print("DATABRICKS API DEPLOYMENT READINESS")
print("=" * 70)

checks = {
    "API function": callable(
        globals().get(
            "build_api_response"
        )
    ),

    "Request validation": callable(
        globals().get(
            "validate_api_request"
        )
    ),

    "Request accepted":
        request_validation.get(
            "valid",
            False
        ),

    "API execution":
        api_response.get(
            "success",
            False
        ),

    "Response structure":
        len(missing_fields) == 0,

    "Valid route":
        actual_route in valid_routes
}

passed = 0

for check_name, status in checks.items():

    print(
        f"{'PASS' if status else 'FAIL'} - "
        f"{check_name}"
    )

    if status:
        passed += 1

failed = (
    len(checks) - passed
)

print()
print(
    "Total checks:",
    len(checks)
)

print(
    "Passed:",
    passed
)

print(
    "Failed:",
    failed
)

if failed:

    raise RuntimeError(
        "Deployment readiness failed."
    )

print()
print(
    "API DEPLOYMENT READINESS: PASS ✓"
)
```
