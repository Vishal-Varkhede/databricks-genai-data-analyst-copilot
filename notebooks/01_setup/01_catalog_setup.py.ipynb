```python
import sys
import pyspark

print("=" * 70)
print("DATABRICKS GENAI DATA ANALYST COPILOT")
print("ENVIRONMENT VERIFICATION")
print("=" * 70)

print(f"Python version : {sys.version.split()[0]}")
print(f"PySpark version: {pyspark.__version__}")
```

```python
print("=" * 70)
print("SPARK VERIFICATION")
print("=" * 70)

spark.range(5).show()
```

```python
print("=" * 70)
print("UNITY CATALOG VERIFICATION")
print("=" * 70)

catalog = spark.sql("SELECT CURRENT_CATALOG()").first()[0]
schema = spark.sql("SELECT CURRENT_SCHEMA()").first()[0]
metastore = spark.sql("SELECT CURRENT_METASTORE()").first()[0]

print("Catalog  :", catalog)
print("Schema   :", schema)
print("Metastore:", metastore)
```

```python
print("=" * 70)
print("AVAILABLE CATALOGS")
print("=" * 70)

spark.sql("SHOW CATALOGS").show(truncate=False)
```

```python
print("=" * 70)
print("AVAILABLE SCHEMAS")
print("=" * 70)

spark.sql(
    f"SHOW SCHEMAS IN `{catalog}`"
).show(truncate=False)
```

```python
print("=" * 70)
print("SPARK SQL VERIFICATION")
print("=" * 70)

spark.sql("""
SELECT
    1 AS verification_id,
    'Databricks GenAI Data Analyst Copilot' AS project,
    current_timestamp() AS verification_time
""").show(truncate=False)
```

```python
print("=" * 70)
print("DELTA VERIFICATION")
print("=" * 70)

test_table = "delta_environment_test"

spark.range(5).write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable(test_table)

print(f"Created Delta table: {test_table}")

spark.table(test_table).show()
```

```python
print("=" * 70)
print("DELTA CLEANUP")
print("=" * 70)

spark.sql(f"DROP TABLE IF EXISTS `{test_table}`")

print("Temporary Delta table removed.")
```

```python
print("=" * 70)
print("ENVIRONMENT VERIFICATION COMPLETE")
print("=" * 70)

print(f"Catalog   : {catalog}")
print(f"Schema    : {schema}")
print(f"Metastore : {metastore}")
print(f"PySpark   : {pyspark.__version__}")

print("\nReady for Phase 2.")
```
